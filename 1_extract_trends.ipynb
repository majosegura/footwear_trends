```python
from pytrends.request import TrendReq
import pandas as pd
import time

# Conexión con Google Trends para México
pytrends = TrendReq(hl='es-MX', tz=360)

# Lista de calzado representativo (máximo 5 por bloque en Trends)
terminos_calzado = [
    "botas vaqueras",
    "mocasines",
    "tenis plataforma",
    "botas chelsea",
    "sandalias plataforma"
]

print("Consultando datos en Google Trends para México...")
pytrends.build_payload(kw_list=terminos_calzado, timeframe='today 5-y', geo='MX')

# Descargar serie de tiempo
df = pytrends.interest_over_time()

```

```python
df
```

```python
# Limpiar columna técnica si existe
if 'isPartial' in df.columns:
    df = df.drop(columns=['isPartial'])

# Guardar en CSV
nombre_archivo = "tendencias_calzado_mexico.csv"
df.to_csv(nombre_archivo)

print(f"¡Listo! Datos guardados en '{nombre_archivo}'.")
print("\nÚltimas 5 semanas de datos:")
print(df.tail())
```
